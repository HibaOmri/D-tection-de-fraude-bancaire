## 1. Importation des Modules et Chargement des Modèles

```python
import sys
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Ajout du dossier racine pour charger nos scripts src
sys.path.insert(0, os.path.abspath(".."))

from src.models_sota import train_and_compare_sota_models
print("Modules importés avec succès !")
```

## 2. Exécution de la Comparaison des Modèles SOTA (Random Forest, XGBoost, LightGBM, CatBoost)
Cette étape entraîne/évalue les 4 modèles ajustés sur le split de validation interne (20% de `fraudTrain.csv`, soit 259 335 transactions).

```python
# Exécution de l'évaluation comparative complète avec matrice de coût bancaire
train_path = "../data/fraudTrain.csv" if os.path.exists("../data/fraudTrain.csv") else "data/fraudTrain.csv"
results = train_and_compare_sota_models(train_path=train_path, models_dir="../models")
```

## 3. Tableau Comparatif Officiel pour le Rapport de Stage
Le tableau ci-dessous regroupe l'ensemble des métriques clés et de la rentabilité financière en Dirhams Marocains (MAD).

```python
df_res = pd.DataFrame([{
    'Modèle': r['model_name'],
    'PR-AUC (Métrique Métier)': round(r['pr_auc'], 4),
    'ROC-AUC': round(r['roc_auc'], 4),
    'Rappel (Recall %)': f"{round(r['recall'] * 100, 2)}%",
    'F1-Score': round(r['f1_score'], 4),
    'Seuil Optimal': round(r['best_threshold'], 4),
    'Économies (MAD / DH)': f"{r['savings_mad']:,.2f} DH",
    'Pertes Évitées (%)': f"{round(r['savings_pct'], 2)}%"
} for r in results])

print("=== TABLEAU COMPARATIF FINAL À INCLURE DANS LE RAPPORT ===")
print(df_res.to_string(index=False))
display(df_res)
```

## 4. Visualisation Graphique des Économies Financières en Dirhams (MAD)
Ce graphique compare directement le montant net préservé par la banque pour chaque modèle.

```python
models_name = [r['model_name'] for r in results]
savings_vals = [r['savings_mad'] for r in results]

plt.figure(figsize=(10, 5))
bars = plt.bar(models_name, [s/1000 for s in savings_vals], color=['#2b5c8f', '#0284c7', '#38bdf8', '#0369a1'])
plt.title("Impact Financier : Économies Nettes pour la Banque (en Milliers de DH)", fontsize=13, fontweight='bold')
plt.ylabel("Économies (en kDH)")
plt.grid(axis='y', linestyle='--', alpha=0.7)

for bar, val in zip(bars, savings_vals):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 10, f"{val:,.0f} DH", ha='center', fontweight='bold')

plt.tight_layout()
plt.show()
```
